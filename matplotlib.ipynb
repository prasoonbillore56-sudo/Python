```python
students=["a","b","c","d","e"]
marks=[10,30,50,20,60]

plt.bar(students,marks,color="purple")
plt.title("student marks")
plt.xlabel("students")
plt.ylabel("marks")
plt.show()
```

```python
data = {
    'Week': [1, 1, 1, 2, 2, 2, 3, 3, 3, 4, 4, 4],
    'Student': ['Alex', 'Emma', 'Ryan', 'Alex', 'Emma', 'Ryan', 'Alex', 'Emma', 'Ryan', 'Alex', 'Emma', 'Ryan'],
    'Study_Hours': [10, 8, 12, 12, 9, 14, 8, 11, 10, 15, 12, 13],
    'Revisions': [2, 1, 3, 3, 2, 4, 1, 3, 2, 4, 3, 3]
}

df = pd.DataFrame(data)

plt.bar(df['Student'], df['Study_Hours'])
plt.xlabel("Student")
plt.ylabel("Study Hours")
plt.title("Study Hours by Students")
plt.show()
```
